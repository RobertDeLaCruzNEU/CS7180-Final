```python
from PIL import Image

import pandas as pd
import matplotlib.pyplot as plt
```

```python
boxes = pd.read_csv("C:/Users/rober/OneDrive/Desktop/CS7150/CS7180-Final/data/annotations/bboxes.csv")
```

```python
boxes.head()
```

```python
boxes.loc[:, 'width']  = boxes.x2 - boxes.x1
boxes.loc[:, 'height'] = boxes.y2 - boxes.y1
```

```python
from PIL import Image
import os
from tqdm import tqdm

groups = boxes.groupby(['video', 'frame'])[['id', 'content', 'x1', 'y1', 'x2', 'y2']].apply(lambda x: x.values.tolist())
for file, xyxy in tqdm(groups.items(), desc="Processing: "):
    video, frame = file
    frame = str(frame).zfill(5)
    path = f"C:/Users/rober/OneDrive/Desktop/CS7150/CS7180-Final/data/images/set01/{video}/{frame}.png"
    img = Image.open(path)
    
    for id, content, x1, y1, x2, y2 in xyxy:
        w = x2 - x1
        h = y2 - y1
        
        # Find the center of the box
        x = x1 + w / 2
        y = y1 + h / 2
        
        # Use the larger dimension to create a square
        side = max(w, h)
        
        # Define new square coordinates
        x1 = max(0, x - side / 2)
        y1 = max(0, y - side / 2)
        x2 = min(img.width, x + side / 2)
        y2 = min(img.height, y + side / 2)
        
        crop = img.crop((x1, y1, x2, y2))
        crop = crop.resize((224, 224), Image.LANCZOS)
        crop.save(f"C:/Users/rober/OneDrive/Desktop/CS7150/CS7180-Final/data/images/cropped/{video}-{frame}-{content}-{int(id)}.png")
```
